This section is defining the dataset

```python
# Define a list of possible responses for the chatbot
greetings = ['Hello!', 'Hi there!', 'Greetings!', 'Hey!', 'Hello! How can I assist you today?', 'Hey! Ready to chat?']
farewells = ['Goodbye!', 'Bye!', 'See you later!', 'Take care!','Take it easy! If you ever need assistance, dont hesitate to return']
help_responses = ['How can I help you?', 'What can I assist you with?', 'What do you need help with?']
unknown_responses = ['I am sorry, I do not understand. Could you please rephrase your question?', 'I am not programmed to answer that.','It seems I may not have the information youre looking for. Can you clarify?',
'I am not quite familiar with that. Could you rephrase or ask another question?',
'I may not have the answer to that. Can you ask something different?',
'I am afraid I do not have information on that particular topic. What else can I assist you with?']
```

```python
# Define a dictionary of possible questions and their corresponding answers
faq = {
    'What is your return policy?': 'We offer a 30-day return policy on most items.', 'Any time!'
    'How can I track my order?': 'You can track your order by logging into your account and checking the order status.',
    'Do you offer international shipping?': 'Yes, we offer international shipping to most countries. Shipping fees may apply.',
    'What payment methods do you accept?': 'We accept all major credit cards, as well as PayPal and Apple Pay.',
}
```

```python
# Define a function to handle user input and provide responses
def chatbot():
    print("Hello, I'm a customer service chatbot. How can I assist you today?")

    while True:
        user_input = input('> ').lower()

        if user_input in ['hi', 'hello', 'hey', 'greetings']:
            print(random.choice(greetings))
        elif user_input in ['bye', 'goodbye', 'see you later']:
            print(random.choice(farewells))
            break
        elif user_input in ['help', 'support']:
            print(random.choice(help_responses))
        else:
            found_answer = False
            for question in faq:
                if user_input in question.lower():
                    print(faq[question])
                    found_answer = True
                    break
            if not found_answer:
                print(random.choice(unknown_responses))
```

```python
# Call the chatbot function to start the conversation
chatbot()
```
